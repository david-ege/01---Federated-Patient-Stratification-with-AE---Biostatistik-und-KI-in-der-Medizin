```python
import pandas as pd
import re

```

```python
df = pd.read_csv('JHU/proteinGroups.txt', delimiter='\t')
```

```python
reporter_columns = list()
rename = list()
for j in range(1,14):
    if j < 10:
        reporter_columns += ['Reporter intensity corrected {} 0{}CPTAC'.format(i, j) for i in range(1, 11)]
        rename += ['0{}CPTAC_{}'.format(j, i) for i in range(1, 11)]
    else :
        reporter_columns += ['Reporter intensity corrected {} {}CPTAC'.format(i, j) for i in range(1, 11)]
        rename += ['{}CPTAC_{}'.format(j, i) for i in range(1, 11)]


export_df = df[reporter_columns]
export_df = export_df.rename(columns=dict(zip(reporter_columns, rename)))

```

```python
meta_df = pd.read_csv('JHU/PCD_study_experimental_CPTAC_info.tsv', delimiter='\t')
meta_df = meta_df.drop(columns=['Run Metadata ID', 'Number of Fractions', 'Protocol'])
```

```python
meta_df
```

```python
def patient_string_processing(patient_string):
    sample, condition = patient_string.split(",")
    if re.search('^Ovarian.*', sample):
        return "Ovarian Tumor Sample"
    else:
        return condition
patient_conditions = meta_df.stack().tolist()
conditions = [patient_string_processing(patient) for patient in patient_conditions]
conditions
```

```python
export_df = pd.concat([export_df, pd.DataFrame( dict(zip(rename, conditions)), index=['Conditions'])], axis=0 )
export_df
```

```python
export_df.to_csv('real_dataset.tsv', sep='\t')
```
