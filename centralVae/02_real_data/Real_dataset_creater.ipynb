```python
import pandas as pd
import re
import numpy as np

```

# Import raw data

```python
df = pd.read_csv('JHU/proteinGroups.txt', delimiter='\t')
```

```python
reporter_columns = list()
rename = list()
for j in range(1,14):
    if j < 10:
        reporter_columns += ['Reporter intensity corrected {} 0{}CPTAC'.format(i, j) for i in range(1, 11)]
        rename += ['0{}CPTAC_{}'.format(j, i) for i in range(1, 11)]
    else :
        reporter_columns += ['Reporter intensity corrected {} {}CPTAC'.format(i, j) for i in range(1, 11)]
        rename += ['{}CPTAC_{}'.format(j, i) for i in range(1, 11)]


export_df = df[reporter_columns]
export_df = export_df.rename(columns=dict(zip(reporter_columns, rename)))
export_df
```

# Normalization

```python
#normalisation skipped for now discussion in meeting
export_df = export_df.apply(lambda x : np.log1p(x))
mean_features = export_df.mean(axis=1)
std_features = export_df.std(axis=1)

export_df_normalized = export_df.apply(lambda x : x - mean_features)
export_df_normalized = export_df_normalized.apply(lambda x : x/std_features)

export_df_normalized = export_df_normalized.fillna(0.0)

def normalized_by_CPTAC(sample_data):
    _, columns = sample_data.shape
    print(columns)
    sample_data_group_normalized = list()
    for i in range(0, columns, 10):
        cptac_samples = sample_data.iloc[:, i: i+10]

        cptac_mean = cptac_samples.mean(axis=1)
        cptac_std = cptac_samples.std(axis=1)

        cptac_samples = cptac_samples.apply(lambda x : x - cptac_mean)
        cptac_samples = cptac_samples.apply(lambda x : x / cptac_std)

        cptac_samples = cptac_samples.fillna(0.0)

        sample_data_group_normalized.append(cptac_samples)
    
    return pd.concat(sample_data_group_normalized, axis=1)

export_df_group_normalized = normalized_by_CPTAC(export_df)
```

# Metadata

```python
meta_df = pd.read_csv('JHU/PCD_study_experimental_CPTAC_info.tsv', delimiter='\t')
meta_df = meta_df.drop(columns=['Run Metadata ID', 'Number of Fractions', 'Protocol'])
```

```python
meta_df
```

```python
def patient_string_processing(patient_string):
    sample, condition = patient_string.split(",")
    if re.search('^Ovarian.*', sample):
        return "Ovarian Tumor Sample"
    else:
        return condition
patient_conditions = meta_df.stack().tolist()
conditions = [patient_string_processing(patient) for patient in patient_conditions]
conditions
```

# Export

```python
condition_df = pd.DataFrame( dict(zip(rename, conditions)), index=['Conditions'])

export_df = pd.concat([export_df, condition_df], axis=0 )
export_df_normalized = pd.concat([export_df_normalized, condition_df], axis=0 )
export_df_group_normalized = pd.concat([export_df_group_normalized, condition_df], axis=0 )

```

```python
export_df.to_csv('real_dataset_raw.tsv', sep='\t')
export_df_normalized.to_csv('real_dataset_normalized.tsv', sep='\t')
export_df_group_normalized.to_csv('real_dataset_normalized_by_cptac.tsv', sep='\t')
```
