# Read and split normalized dataset

```python
df = pd.read_csv('data/real_dataset_normalized_by_cptac.tsv', sep='\t', index_col=0)
```

```python
clients = 7

# array distributing samples to clients 
nr_batches = 13
base_value =  nr_batches // clients
remainder = nr_batches % clients 
cptac_distr = [base_value] * clients  
for i in range(remainder):
    cptac_distr[i] += 1 

print(cptac_distr)
start = 0
up_to = 0
for client in range(clients):
    up_to += cptac_distr[client] * 10
    client += 1
    print(start)
    print(up_to)
    part = df.iloc[:, start:up_to].T
    part_conditions = part.iloc[:, -1]
    part.drop(part.columns[len(part.columns)-1], axis=1, inplace=True)
    part = part.reset_index(drop=True)
    part.to_csv(f'data/client{client}/data/realData.csv', sep=';')
    part_conditions.to_csv(f'data/client{client}/data/realmetadata.csv', sep=';')
    start = up_to
    print(part)
```
